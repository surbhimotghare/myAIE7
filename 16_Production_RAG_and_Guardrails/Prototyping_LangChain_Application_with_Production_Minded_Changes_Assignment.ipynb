#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

##### ❓ Question #1: Production Caching Analysis

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.

#### ✅ Answer:

##### **Limitations of Current Caching Approach and Trade-offs:**

**1. Memory vs Disk Trade-offs**
- **Memory caching**: Fast but lost on restart, limited by RAM
- **File-based caching**: Persistent but slower I/O, can grow indefinitely
- **Current approach**: Uses file storage which can consume significant disk space

**2. Cache Invalidation Issues**
- **No automatic expiration**: Cached embeddings never expire, even if source documents change
- **Model updates**: Cache becomes invalid when switching embedding models
- **Stale data**: No mechanism to detect when PDF content has been updated

**3. Concurrent Access Problems**
- **File locking**: Multiple processes accessing same cache files can cause conflicts
- **Race conditions**: Simultaneous writes to cache directory
- **No coordination**: Each process maintains separate cache state

**4. Cache Size Management**
- **Unlimited growth**: No size limits or cleanup policies
- **Storage bloat**: Large documents create massive cache directories
- **No LRU eviction**: Old, unused embeddings never get removed

**5. Cold Start Scenarios**
- **Empty cache penalty**: First-time users experience full latency
- **Deployment resets**: Cache lost during container restarts
- **Scaling issues**: Each new instance starts with cold cache

##### **Most/Least Useful Scenarios:**

**Most Useful:**
- Repeated queries on same documents
- Development/testing environments
- Single-user applications
- Static document sets

**Least Useful:**
- Frequently changing documents
- High-concurrency environments
- Memory-constrained systems
- One-time query patterns

**Production Improvements Needed:**
- Redis/database caching
- Cache expiration policies
- Distributed cache coordination
- Size limits and cleanup
- Semantic similarity caching

I added comprehensive cache performance tests below to evaluate:

These tests have key features as follows:

````python
### YOUR CODE HERE

"""
Activity #1: Cache Performance Testing
"""

import time
import statistics
from typing import List, Dict

def test_cache_performance(rag_chain, num_tests: int = 5):
    """
    Comprehensive cache performance testing for the production RAG system.
    
    Tests:
    1. Embedding cache performance
    2. LLM cache performance  
    3. Cache hit rate measurements
    4. Performance variance analysis
    """
    
    print("🧪 Starting Cache Performance Testing...")
    print("=" * 60)
    
    # Test queries - mix of similar and different queries
    test_queries = [
        "What is this document about?",
        "What are the main topics covered?",
        "What is this document about?",  # Duplicate for cache testing
        "What are the key requirements?",
        "What is this document about?",  # Another duplicate
    ]
    
    results = {
        'first_calls': [],
        'cache_hits': [],
        'queries': [],
        'cache_hit_rate': 0
    }
    
    print(f"📊 Running {len(test_queries)} queries with {num_tests} repetitions each...\n")
    
    for i, query in enumerate(test_queries, 1):
        print(f"🔍 Test {i}: '{query[:50]}{'...' if len(query) > 50 else ''}'")
        
        # Track if this is a repeated query
        is_repeated = query in results['queries']
        results['queries'].append(query)
        
        query_times = []
        
        for test_num in range(num_tests):
            start_time = time.time()
            
            try:
                response = rag_chain.invoke(query)
                end_time = time.time()
                
                elapsed = end_time - start_time
                query_times.append(elapsed)
                
                # Show first response preview
                if test_num == 0:
                    preview = response.content[:100] + "..." if len(response.content) > 100 else response.content
                    print(f"   Response preview: {preview}")
                
            except Exception as e:
                print(f"   ❌ Error in test {test_num + 1}: {e}")
                continue
        
        if query_times:
            avg_time = statistics.mean(query_times)
            min_time = min(query_times)
            max_time = max(query_times)
            std_dev = statistics.stdev(query_times) if len(query_times) > 1 else 0
            
            print(f"   ⏱️  Average time: {avg_time:.3f}s")
            print(f"   📈 Min/Max: {min_time:.3f}s / {max_time:.3f}s")
            print(f"   📊 Std dev: {std_dev:.3f}s")
            
            # Categorize as cache hit or miss based on whether it's repeated
            if is_repeated:
                results['cache_hits'].extend(query_times)
                print(f"   💾 Cache hit detected (repeated query)")
            else:
                results['first_calls'].extend(query_times)
                print(f"   🆕 First call (cache miss)")
                
        print()
    
    # Calculate overall statistics
    if results['first_calls'] and results['cache_hits']:
        avg_first_call = statistics.mean(results['first_calls'])
        avg_cache_hit = statistics.mean(results['cache_hits'])
        speedup = avg_first_call / avg_cache_hit if avg_cache_hit > 0 else float('inf')
        
        cache_hit_count = len([q for q in results['queries'] if results['queries'].count(q) > 1])
        total_queries = len(results['queries'])
        cache_hit_rate = (cache_hit_count / total_queries) * 100
        
        print("📈 PERFORMANCE SUMMARY:")
        print("=" * 40)
        print(f"🔄 Average first call time: {avg_first_call:.3f}s")
        print(f"⚡ Average cache hit time:  {avg_cache_hit:.3f}s")
        print(f"🚀 Cache speedup:          {speedup:.1f}x faster")
        print(f"💾 Cache hit rate:         {cache_hit_rate:.1f}%")
        print(f"📊 Total queries tested:   {total_queries}")
        
        # Performance insights
        print("\n🔍 PERFORMANCE INSIGHTS:")
        print("=" * 40)
        
        if speedup > 5:
            print("✅ Excellent cache performance - significant speedup detected")
        elif speedup > 2:
            print("✅ Good cache performance - noticeable speedup")
        else:
            print("⚠️  Limited cache benefit - consider optimizations")
            
        if cache_hit_rate > 40:
            print("✅ High cache utilization - good for production workloads")
        else:
            print("⚠️  Low cache utilization - mostly unique queries")
    
    return results


def test_embedding_cache_directly(rag_chain, test_text: str = "Sample text for embedding cache test"):
    """
    Test embedding cache performance directly by accessing the cached embeddings.
    """
    print("\n🔬 EMBEDDING CACHE DIRECT TEST:")
    print("=" * 40)
    
    try:
        # Get the cached embeddings object
        cached_embeddings = rag_chain.cached_embeddings.get_embeddings()
        
        print(f"📝 Test text: '{test_text}'")
        
        # First embedding call (cache miss)
        print("\n1️⃣ First embedding call (cache miss):")
        start_time = time.time()
        embedding1 = cached_embeddings.embed_query(test_text)
        first_time = time.time() - start_time
        print(f"   ⏱️  Time: {first_time:.3f}s")
        print(f"   📏 Embedding dimensions: {len(embedding1)}")
        
        # Second embedding call (cache hit)
        print("\n2️⃣ Second embedding call (cache hit):")
        start_time = time.time()
        embedding2 = cached_embeddings.embed_query(test_text)
        second_time = time.time() - start_time
        print(f"   ⏱️  Time: {second_time:.3f}s")
        print(f"   📏 Embedding dimensions: {len(embedding2)}")
        
        # Verify embeddings are identical
        embeddings_identical = embedding1 == embedding2
        print(f"   🔍 Embeddings identical: {embeddings_identical}")
        
        if first_time > 0 and second_time > 0:
            speedup = first_time / second_time
            print(f"\n🚀 Embedding cache speedup: {speedup:.1f}x faster")
            
            if speedup > 10:
                print("✅ Excellent embedding cache performance")
            elif speedup > 3:
                print("✅ Good embedding cache performance")
            else:
                print("⚠️  Limited embedding cache benefit")
        
    except Exception as e:
        print(f"❌ Error testing embedding cache: {e}")


def test_llm_cache_variations(rag_chain):
    """
    Test LLM cache with slight variations in queries to understand cache sensitivity.
    """
    print("\n🎯 LLM CACHE VARIATION TEST:")
    print("=" * 40)
    
    # Test queries with slight variations
    query_variations = [
        "What is this document about?",
        "What is this document about?",  # Exact duplicate
        "What is this document about ",  # Extra space
        "what is this document about?",  # Different case
        "What is this document about??", # Extra punctuation
        "What's this document about?",   # Contraction
    ]
    
    results = []
    
    for i, query in enumerate(query_variations, 1):
        print(f"\n{i}️⃣ Query: '{query}'")
        
        start_time = time.time()
        try:
            response = rag_chain.invoke(query)
            elapsed = time.time() - start_time
            
            results.append({
                'query': query,
                'time': elapsed,
                'response_length': len(response.content)
            })
            
            print(f"   ⏱️  Time: {elapsed:.3f}s")
            print(f"   📝 Response length: {len(response.content)} chars")
            
            # Detect potential cache hit (very fast response)
            if elapsed < 0.5:
                print("   💾 Likely cache hit")
            else:
                print("   🔄 Likely cache miss")
                
        except Exception as e:
            print(f"   ❌ Error: {e}")
    
    # Analyze results
    if len(results) > 1:
        times = [r['time'] for r in results]
        avg_time = statistics.mean(times)
        min_time = min(times)
        max_time = max(times)
        
        print(f"\n📊 VARIATION TEST SUMMARY:")
        print(f"   Average time: {avg_time:.3f}s")
        print(f"   Fastest: {min_time:.3f}s")
        print(f"   Slowest: {max_time:.3f}s")
        print(f"   Time variance: {max_time - min_time:.3f}s")


# Main test runner
def run_all_cache_tests(rag_chain):
    """
    Run all cache performance tests.
    """
    print("🚀 COMPREHENSIVE CACHE PERFORMANCE TESTING")
    print("=" * 60)
    
    # Test 1: Overall cache performance
    overall_results = test_cache_performance(rag_chain, num_tests=3)
    
    # Test 2: Direct embedding cache test
    test_embedding_cache_directly(rag_chain)
    
    # Test 3: LLM cache variation sensitivity
    test_llm_cache_variations(rag_chain)
    
    print("\n" + "=" * 60)
    print("✅ All cache performance tests completed!")
    print("💡 Use these insights to optimize your production caching strategy.")
    
    return overall_results
````

#### ✅ Answer: Create and Test Helpfulness Agent

I created a Helpfulness Agent and it utilizes the cached RAG chain created earlier in the notebook

````python
# Helpfulness Agent Implementation 

# Import all necessary components
from typing import Optional, List, Dict, Any
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import AIMessage, HumanMessage
from langgraph_agent_lib.agents import get_default_tools, AgentState

def create_helpfulness_agent(
    model_name: str = "gpt-4o-mini",
    temperature: float = 0.1,
    tools: Optional[List] = None,
    rag_chain: Optional[ProductionRAGChain] = None,
    max_iterations: int = 3
):
    """Create a LangGraph agent with helpfulness evaluation and refinement.
    
    Args:
        model_name: OpenAI model name
        temperature: Model temperature
        tools: List of tools to bind to the model
        rag_chain: Optional RAG chain to include as a tool
        max_iterations: Maximum number of refinement iterations
        
    Returns:
        Compiled LangGraph agent with helpfulness checking
    """
    if tools is None:
        tools = get_default_tools(rag_chain)
    
    # Get model and bind tools
    model = get_openai_model(model_name=model_name, temperature=temperature)
    model_with_tools = model.bind_tools(tools)
    
    def call_model(state: AgentState) -> Dict[str, Any]:
        """Invoke the model with messages and append response."""
        messages = state["messages"]
        response = model_with_tools.invoke(messages)
        return {"messages": [response]}
    
    def route_to_action_or_helpfulness(state: AgentState):
        """Decide whether to execute tools or run helpfulness evaluator."""
        last_message = state["messages"][-1]
        if getattr(last_message, "tool_calls", None):
            return "action"
        return "helpfulness"
    
    def helpfulness_node(state: AgentState) -> Dict[str, Any]:
        """Evaluate helpfulness of the latest response relative to initial query."""
        # Prevent infinite loops by limiting iterations
        helpfulness_messages = [m for m in state["messages"] if 
                              hasattr(m, 'content') and 'HELPFULNESS:' in str(m.content)]
        
        if len(helpfulness_messages) >= max_iterations:
            return {"messages": [AIMessage(content="HELPFULNESS:END")]}
        
        # Get initial query and final response
        initial_query = state["messages"][0]
        final_response = state["messages"][-1]
        
        prompt_template = """
Given an initial query and a final response, determine if the final response is extremely helpful or not. 

Consider these criteria:
- Does it directly address the user's question?
- Is the information accurate and relevant?
- Is it comprehensive enough?
- Is it clear and well-structured?

Please indicate helpfulness with 'Y' for helpful responses and 'N' for responses that need improvement.

Initial Query:
{initial_query}

Final Response:
{final_response}

Helpfulness Assessment:"""

        helpfulness_prompt = PromptTemplate.from_template(prompt_template)
        helpfulness_model = get_openai_model(model_name="gpt-4o-mini", temperature=0.0)
        helpfulness_chain = helpfulness_prompt | helpfulness_model | StrOutputParser()
        
        try:
            helpfulness_response = helpfulness_chain.invoke({
                "initial_query": initial_query.content,
                "final_response": final_response.content,
            })
            
            # Determine decision based on response
            decision = "Y" if "Y" in helpfulness_response.upper() else "N"
            return {"messages": [AIMessage(content=f"HELPFULNESS:{decision}")]}
            
        except Exception as e:
            # If evaluation fails, assume helpful to avoid loops
            print(f"⚠ Helpfulness evaluation error: {e}")
            return {"messages": [AIMessage(content="HELPFULNESS:Y")]}
    
    def helpfulness_decision(state: AgentState):
        """Route based on helpfulness evaluation."""
        # Check for loop limit
        if any(getattr(m, "content", "") == "HELPFULNESS:END" for m in state["messages"][-1:]):
            return END
        
        last_message = state["messages"][-1]
        content = getattr(last_message, "content", "")
        
        if "HELPFULNESS:Y" in content:
            return END  # Response is helpful, end conversation
        elif "HELPFULNESS:N" in content:
            return "agent"  # Response needs improvement, continue refining
        else:
            return END  # Fallback to end
    
    # Build the graph
    graph = StateGraph(AgentState)
    tool_node = ToolNode(tools)
    
    # Add nodes
    graph.add_node("agent", call_model)
    graph.add_node("action", tool_node)
    graph.add_node("helpfulness", helpfulness_node)
    
    # Set entry point
    graph.set_entry_point("agent")
    
    # Add conditional edges
    graph.add_conditional_edges(
        "agent",
        route_to_action_or_helpfulness,
        {"action": "action", "helpfulness": "helpfulness"}
    )
    
    graph.add_conditional_edges(
        "helpfulness",
        helpfulness_decision,
        {"agent": "agent", END: END}
    )
    
    graph.add_edge("action", "agent")
    
    return graph.compile()


# Create and Test Helpfulness Agent
print("Creating Helpfulness Agent with RAG integration...")

try:
    helpfulness_agent = create_helpfulness_agent(
        model_name="gpt-4o-mini",
        temperature=0.1,
        rag_chain=rag_chain,  # Pass our cached RAG chain
        max_iterations=3
    )
    print("✓ Helpfulness Agent created successfully!")
    print("  - Model: gpt-4o-mini")
    print("  - Tools: Tavily Search, Arxiv, RAG System")
    print("  - Features: Tool calling, helpfulness evaluation, iterative refinement")
    print("  - Max refinement iterations: 3")
    
except Exception as e:
    print(f"❌ Error creating helpfulness agent: {e}")
    helpfulness_agent = None


# Test Helpfulness Agent
print("\n🤖 Testing Helpfulness Agent...")
print("=" * 50)

test_query_helpfulness = "What are the key benefits and requirements for Direct Loan forgiveness programs?"

if helpfulness_agent:
    try:        
        # Create message for the agent
        messages = [HumanMessage(content=test_query_helpfulness)]
        
        print(f"Query: {test_query_helpfulness}")
        print("\n🔄 Helpfulness Agent Response (with evaluation):")
        
        # Invoke the agent
        response = helpfulness_agent.invoke({"messages": messages})
        
        # Extract the final message (skip helpfulness evaluation messages)
        final_messages = [msg for msg in response["messages"] 
                         if not (hasattr(msg, 'content') and 'HELPFULNESS:' in str(msg.content))]
        
        if final_messages:
            final_message = final_messages[-1]
            print(final_message.content)
        
        # Show conversation analysis
        total_messages = len(response["messages"])
        helpfulness_checks = len([m for m in response["messages"] 
                                if hasattr(m, 'content') and 'HELPFULNESS:' in str(m.content)])
        
        print(f"\n📊 Conversation Analysis:")
        print(f"   Total messages: {total_messages}")
        print(f"   Helpfulness evaluations: {helpfulness_checks}")
        print(f"   Refinement iterations: {helpfulness_checks}")
        
        # Show if the response was refined
        if helpfulness_checks > 1:
            print("   🔄 Response was refined through helpfulness evaluation")
        else:
            print("   ✅ Response was deemed helpful on first attempt")
        
    except Exception as e:
        print(f"❌ Error testing helpfulness agent: {e}")
else:
    print("⚠ Helpfulness agent not available - skipping test")


# Compare Simple vs Helpfulness Agent
print("\n🔍 Agent Comparison Test")
print("=" * 50)

comparison_query = "How do I apply for student loan deferment?"

print(f"Query: {comparison_query}")

# Test Simple Agent
if 'simple_agent' in locals() and simple_agent:
    print("\n📝 Simple Agent Response:")
    try:
        simple_response = simple_agent.invoke({"messages": [HumanMessage(content=comparison_query)]})
        simple_final = simple_response["messages"][-1]
        print(f"Length: {len(simple_final.content)} characters")
        print(f"Preview: {simple_final.content[:200]}...")
        simple_message_count = len(simple_response["messages"])
    except Exception as e:
        print(f"❌ Simple agent error: {e}")
        simple_message_count = 0

# Test Helpfulness Agent
if helpfulness_agent:
    print("\n🎯 Helpfulness Agent Response:")
    try:
        help_response = helpfulness_agent.invoke({"messages": [HumanMessage(content=comparison_query)]})
        help_final_messages = [msg for msg in help_response["messages"] 
                             if not (hasattr(msg, 'content') and 'HELPFULNESS:' in str(msg.content))]
        if help_final_messages:
            help_final = help_final_messages[-1]
            print(f"Length: {len(help_final.content)} characters")
            print(f"Preview: {help_final.content[:200]}...")
        
        help_message_count = len(help_response["messages"])
        help_evaluations = len([m for m in help_response["messages"] 
                              if hasattr(m, 'content') and 'HELPFULNESS:' in str(m.content)])
        
        print(f"\n🔍 Comparison Summary:")
        if 'simple_message_count' in locals():
            print(f"   Simple Agent: {simple_message_count} total messages")
        print(f"   Helpfulness Agent: {help_message_count} total messages ({help_evaluations} evaluations)")
        
        if help_evaluations > 1:
            print("   ✨ Helpfulness agent performed iterative refinement")
        else:
            print("   ✅ Both agents produced satisfactory responses")
            
    except Exception as e:
        print(f"❌ Helpfulness agent error: {e}")

print("\n" + "=" * 50)
print("✅ Agent comparison completed!")
````

````python
# Test different query types

import time
from langchain_core.messages import HumanMessage

def test_different_query_types(simple_agent, helpfulness_agent):
    """Test different types of queries to observe tool selection patterns."""
    
    print("🧪 TESTING DIFFERENT QUERY TYPES")
    print("=" * 50)
    
    # Test queries categorized by expected tool usage
    test_cases = [
        {
            "category": "RAG-focused (Document Knowledge)",
            "queries": [
                "What is the main purpose of the Direct Loan Program?",
                "What are the eligibility requirements mentioned in this document?",
                "How long is the grace period for student loans?"
            ],
            "expected_tool": "RAG System"
        },
        {
            "category": "Web Search (Current Events)", 
            "queries": [
                "What are the latest developments in AI safety?",
                "What are the current federal student loan interest rates for 2024?",
                "What recent changes have been made to student loan policies?"
            ],
            "expected_tool": "Tavily Search"
        },
        {
            "category": "Academic Search (Research)",
            "queries": [
                "Find recent papers about transformer architectures",
                "What research exists on student loan debt impact?",
                "Show me studies on educational financing effectiveness"
            ],
            "expected_tool": "Arxiv"
        },
        {
            "category": "Multi-tool (Complex)",
            "queries": [
                "How do the concepts in this document relate to current AI research trends?",
                "Compare Direct Loan policies with recent federal aid changes",
                "What do recent studies say about loan forgiveness effectiveness?"
            ],
            "expected_tool": "Multiple Tools"
        }
    ]
    
    results = []
    
    for test_case in test_cases:
        print(f"\n📋 {test_case['category']}")
        print(f"Expected tool: {test_case['expected_tool']}")
        print("-" * 40)
        
        for query in test_case['queries'][:2]:  # Test first 2 queries from each category
            print(f"\n🔍 Query: {query}")
            
            # Test Simple Agent
            print("  🤖 Simple Agent:")
            start_time = time.time()
            try:
                simple_response = simple_agent.invoke({"messages": [HumanMessage(content=query)]})
                simple_time = time.time() - start_time
                
                # Extract tools used
                simple_tools = []
                for msg in simple_response["messages"]:
                    if hasattr(msg, 'tool_calls') and msg.tool_calls:
                        for tool_call in msg.tool_calls:
                            simple_tools.append(tool_call['name'])
                
                print(f"    ⏱️  Time: {simple_time:.2f}s")
                print(f"    🔧 Tools: {', '.join(set(simple_tools)) if simple_tools else 'None'}")
                print(f"    📝 Response: {simple_response['messages'][-1].content[:100]}...")
                
            except Exception as e:
                simple_time = time.time() - start_time
                print(f"    ❌ Error ({simple_time:.2f}s): {e}")
                simple_tools = []
            
            # Test Helpfulness Agent  
            print("  🎯 Helpfulness Agent:")
            start_time = time.time()
            try:
                help_response = helpfulness_agent.invoke({"messages": [HumanMessage(content=query)]})
                help_time = time.time() - start_time
                
                # Extract tools used and evaluations
                help_tools = []
                evaluations = 0
                for msg in help_response["messages"]:
                    if hasattr(msg, 'tool_calls') and msg.tool_calls:
                        for tool_call in msg.tool_calls:
                            help_tools.append(tool_call['name'])
                    if hasattr(msg, 'content') and 'HELPFULNESS:' in str(msg.content):
                        evaluations += 1
                
                # Get final response (skip evaluation messages)
                final_messages = [msg for msg in help_response["messages"] 
                                if not (hasattr(msg, 'content') and 'HELPFULNESS:' in str(msg.content))]
                final_content = final_messages[-1].content if final_messages else "No response"
                
                print(f"    ⏱️  Time: {help_time:.2f}s")
                print(f"    🔧 Tools: {', '.join(set(help_tools)) if help_tools else 'None'}")
                print(f"    🔍 Evaluations: {evaluations}")
                print(f"    📝 Response: {final_content[:100]}...")
                
            except Exception as e:
                help_time = time.time() - start_time
                print(f"    ❌ Error ({help_time:.2f}s): {e}")
                help_tools = []
                evaluations = 0
            
            # Store results for analysis
            results.append({
                'category': test_case['category'],
                'query': query,
                'simple_time': simple_time if 'simple_time' in locals() else 0,
                'help_time': help_time if 'help_time' in locals() else 0,
                'simple_tools': simple_tools if 'simple_tools' in locals() else [],
                'help_tools': help_tools if 'help_tools' in locals() else [],
                'evaluations': evaluations if 'evaluations' in locals() else 0
            })
    
    return results

def test_cache_performance(simple_agent, helpfulness_agent):
    """Test cache performance with repeated and similar queries."""
    
    print(f"\n💾 CACHE PERFORMANCE TESTING")
    print("=" * 50)
    
    # Test with repeated queries
    cache_queries = [
        "What is the main purpose of the Direct Loan Program?",  # First time
        "What are the eligibility requirements for Direct Loans?",  # First time  
        "What is the main purpose of the Direct Loan Program?",  # Repeat - should be cached
        "What are the eligibility requirements for Direct Loans?",  # Repeat - should be cached
        "What is the main purpose of the direct loan program?",  # Similar but different case
    ]
    
    print("Testing cache behavior with repeated queries:")
    
    for i, query in enumerate(cache_queries, 1):
        is_repeat = query in cache_queries[:i-1]
        is_similar = any(q.lower() == query.lower() and q != query for q in cache_queries[:i-1])
        
        status = "🔄 REPEAT" if is_repeat else "🔀 SIMILAR" if is_similar else "🆕 NEW"
        print(f"\n{i}. {status}: {query}")
        
        # Test Simple Agent
        start_time = time.time()
        try:
            simple_response = simple_agent.invoke({"messages": [HumanMessage(content=query)]})
            simple_time = time.time() - start_time
            cache_indicator = "⚡ (likely cached)" if simple_time < 0.5 else "🔄 (likely not cached)"
            print(f"  Simple Agent: {simple_time:.2f}s {cache_indicator}")
        except Exception as e:
            print(f"  Simple Agent: ❌ Error: {e}")
        
        # Test Helpfulness Agent
        start_time = time.time()
        try:
            help_response = helpfulness_agent.invoke({"messages": [HumanMessage(content=query)]})
            help_time = time.time() - start_time
            cache_indicator = "⚡ (likely cached)" if help_time < 1.0 else "🔄 (likely not cached)"
            print(f"  Helpfulness Agent: {help_time:.2f}s {cache_indicator}")
        except Exception as e:
            print(f"  Helpfulness Agent: ❌ Error: {e}")

def test_production_readiness(simple_agent, helpfulness_agent):
    """Test production readiness scenarios."""
    
    print(f"\n🛠️ PRODUCTION READINESS TESTING")
    print("=" * 50)
    
    # Test edge cases and error scenarios
    edge_cases = [
        ("Empty query", ""),
        ("Very short query", "Hi"),
        ("Very long query", "What " * 100 + "is this document about?"),
        ("Special characters", "What's the $$$ cost of loans? @#$%"),
        ("Non-English", "¿Cuál es el propósito del programa de préstamos directos?"),
        ("Irrelevant topic", "How do I bake a chocolate cake?"),
    ]
    
    print("Testing edge cases and error handling:")
    
    for case_name, query in edge_cases:
        print(f"\n🧪 {case_name}: '{query[:50]}{'...' if len(query) > 50 else ''}'")
        
        # Test both agents
        for agent, name in [(simple_agent, "Simple"), (helpfulness_agent, "Helpfulness")]:
            start_time = time.time()
            try:
                response = agent.invoke({"messages": [HumanMessage(content=query)]})
                response_time = time.time() - start_time
                
                # Check if we got a reasonable response
                if response and response.get("messages"):
                    final_msg = response["messages"][-1]
                    response_length = len(final_msg.content) if hasattr(final_msg, 'content') else 0
                    print(f"  {name}: ✅ Handled gracefully ({response_time:.2f}s, {response_length} chars)")
                else:
                    print(f"  {name}: ⚠️  No response ({response_time:.2f}s)")
                    
            except Exception as e:
                response_time = time.time() - start_time
                print(f"  {name}: ❌ Error ({response_time:.2f}s): {str(e)[:60]}...")

def analyze_testing_results(results):
    """Analyze and summarize testing results."""
    
    print(f"\n📊 TESTING ANALYSIS & INSIGHTS")
    print("=" * 50)
    
    if not results:
        print("No results to analyze.")
        return
    
    # Performance analysis
    simple_times = [r['simple_time'] for r in results if r['simple_time'] > 0]
    help_times = [r['help_time'] for r in results if r['help_time'] > 0]
    
    if simple_times and help_times:
        avg_simple = sum(simple_times) / len(simple_times)
        avg_help = sum(help_times) / len(help_times)
        
        print(f"⏱️  PERFORMANCE SUMMARY:")
        print(f"  Simple Agent avg: {avg_simple:.2f}s")
        print(f"  Helpfulness Agent avg: {avg_help:.2f}s")
        print(f"  Performance overhead: {avg_help/avg_simple:.1f}x")
    
    # Tool usage analysis
    print(f"\n🔧 TOOL USAGE PATTERNS:")
    categories = {}
    for r in results:
        cat = r['category']
        if cat not in categories:
            categories[cat] = {'simple_tools': [], 'help_tools': []}
        categories[cat]['simple_tools'].extend(r['simple_tools'])
        categories[cat]['help_tools'].extend(r['help_tools'])
    
    for category, tools in categories.items():
        simple_tools = set(tools['simple_tools'])
        help_tools = set(tools['help_tools'])
        print(f"  {category}:")
        print(f"    Simple Agent: {', '.join(simple_tools) if simple_tools else 'No tools'}")
        print(f"    Helpfulness Agent: {', '.join(help_tools) if help_tools else 'No tools'}")
    
    # Quality metrics
    total_evaluations = sum(r['evaluations'] for r in results)
    queries_with_evaluations = len([r for r in results if r['evaluations'] > 0])
    
    print(f"\n🎯 QUALITY METRICS:")
    print(f"  Total helpfulness evaluations: {total_evaluations}")
    print(f"  Queries requiring evaluation: {queries_with_evaluations}/{len(results)}")
    if queries_with_evaluations > 0:
        print(f"  Avg evaluations per query: {total_evaluations/queries_with_evaluations:.1f}")

# Main execution function - replace the commented code with this:
print("🚀 RUNNING ADVANCED AGENT TESTING...")

# Check if agents are available
if 'simple_agent' in locals() and simple_agent and 'helpfulness_agent' in locals() and helpfulness_agent:
    
    # 1. Test different query types and tool selection
    results = test_different_query_types(simple_agent, helpfulness_agent)
    
    # 2. Test cache performance
    test_cache_performance(simple_agent, helpfulness_agent)
    
    # 3. Test production readiness
    test_production_readiness(simple_agent, helpfulness_agent)
    
    # 4. Analyze results
    analyze_testing_results(results)
    
    print(f"\n" + "=" * 60)
    print("✅ ADVANCED AGENT TESTING COMPLETED!")
    print("💡 Review the results above to understand:")
    print("   - Tool selection patterns for different query types")
    print("   - Cache performance and hit rates") 
    print("   - Production readiness and error handling")
    print("   - Performance trade-offs between agent architectures")
    
else:
    print("❌ Agents not available!")
    print("Please ensure both simple_agent and helpfulness_agent are created first.")
    print("Run the previous cells to create the agents before running this test.")
````

Now we'll integrate **Guardrails AI** into our production system to ensure our agents operate safely and within acceptable boundaries. Guardrails provide essential safety layers for production LLM applications by validating inputs, outputs, and behaviors.

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

**🏢 Enterprise Requirements:**
- **Compliance**: Meet regulatory requirements for data protection
- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses
- **Quality Assurance**: Ensure factual accuracy and relevance

**⚡ Technical Advantages:**
- **Layered Defense**: Multiple validation stages for robust protection
- **Selective Enforcement**: Different guards for different use cases
- **Performance Optimization**: Fast validation without sacrificing accuracy
- **Integration Ready**: Works seamlessly with LangGraph agent workflows
